```python
# Definim functia sigmoid
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

# Definim functia MSE
def mean_squared_error(y_pred, y_true):
  return ((y_pred - y_true)**2).sum() / (y_pred.size * 2)

# Definim acuratetea
def accuracy(y_pred, y_true):
  acc = y_pred.argmax(axis=1) == y_true.argmax(axis=1)
  return acc.mean()

# Definim functia one_hot
def one_hot(x, num_classes=output_size):
  result = np.zeros((x.size, num_classes))
  result[np.arange(x.size), x] = 1
  return result
```

# Testarea funcționalităților

Să se antreneze o rețea neuronală simplă (2 straturi ascunse, toate straturile complet conectate) în PyTorch pentru clasificarea datelor din baza de date Wine.

Parametrii de mai jos pot fi modificați pentru a compara comportamentul rețelei:

| Parametru | Variante (implicit **îngroșat**) |
|---|---|
| Funcția de activare | **ReLU**, Sigmoid, Tanh |
| Funcția de cost | **Cross-Entropy**, MSE, Hinge (MultiMargin) |
| Rata de învățare | 0.1, **0.01**, 0.001 |
| Dimensiunea batch-ului | 8, **16**, 142 (tot setul de antrenare) |
| Numărul de epoci | 50, **100**, 300 |
| Optimizatorul | SGD, SGD cu momentum, **Adam** |
| Regularizarea | **niciuna**, L2 (weight decay), Dropout |
| Standardizarea intrărilor | da, **nu** |

Pentru fiecare test, rulați din nou toate celulele de mai jos (începând cu celula de parametri).

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Folosim GPU daca este disponibil
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Rulam pe: {}".format(device))
```

```python
# Alegeti parametrii pentru test (in Colab apar ca liste derulante)
ACTIVARE = "relu"               #@param ["relu", "sigmoid", "tanh"]
FUNCTIE_COST = "cross_entropy"  #@param ["cross_entropy", "mse", "hinge"]
LEARNING_RATE = 0.01            #@param [0.1, 0.01, 0.001] {type:"raw"}
BATCH_SIZE = 16                 #@param [8, 16, 142] {type:"raw"}
NR_EPOCI = 100                  #@param [50, 100, 300] {type:"raw"}
OPTIMIZATOR = "adam"            #@param ["sgd", "sgd_momentum", "adam"]
REGULARIZARE = "niciuna"        #@param ["niciuna", "l2", "dropout"]
STANDARDIZARE = False           #@param {type:"boolean"}

# Dimensiunile straturilor ascunse
HIDDEN_1 = 16
HIDDEN_2 = 16

# Initializam seed-ul generatoarelor aleatoare, pentru ca testele sa fie comparabile
SEED = 0
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# Incarcam baza de date Wine si o impartim in train (80%) vs test (20%), pastrand proportia claselor
data = load_wine()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, stratify=data.target, random_state=SEED)

# Standardizam intrarile folosind media si deviatia standard calculate DOAR pe setul de antrenare
# Fara standardizare, caracteristicile au scari foarte diferite (e.g. 'proline' are valori de ordinul sutelor/miilor)
if STANDARDIZARE:
    mean, std = X_train.mean(axis=0), X_train.std(axis=0)
    X_train = (X_train - mean) / std
    X_test = (X_test - mean) / std

# Transformam datele in tensori PyTorch
X_train = torch.tensor(X_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)

# DataLoader-ul imparte datele in batch-uri si le amesteca la fiecare epoca
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test, y_test), batch_size=len(y_test))

input_size = X_train.shape[1]
output_size = len(data.target_names)
print("Train: {} esantioane, Test: {} esantioane, {} clase".format(len(y_train), len(y_test), output_size))
```

```python
# Definim reteaua: intrare -> FC -> activare -> FC -> activare -> FC (iesire)
class ShallowNet(nn.Module):
    def __init__(self, input_size, hidden_1, hidden_2, output_size, activare, p_dropout=0.0):
        super().__init__()
        activari = {"relu": nn.ReLU, "sigmoid": nn.Sigmoid, "tanh": nn.Tanh}
        self.layers = nn.Sequential(
            nn.Linear(input_size, hidden_1),
            activari[activare](),
            nn.Dropout(p_dropout),
            nn.Linear(hidden_1, hidden_2),
            activari[activare](),
            nn.Dropout(p_dropout),
            nn.Linear(hidden_2, output_size),
        )

    def forward(self, x):
        # Iesirea retelei sunt scorurile nenormalizate (logits) pentru fiecare clasa
        return self.layers(x)

# Dropout-ul este activ doar daca a fost ales ca metoda de regularizare (p=0 il dezactiveaza)
p_dropout = 0.3 if REGULARIZARE == "dropout" else 0.0
model = ShallowNet(input_size, HIDDEN_1, HIDDEN_2, output_size, ACTIVARE, p_dropout).to(device)
print(model)
```

```python
# Definim functia de cost. Toate variantele primesc logits si etichete (indicii claselor).
def mse_loss(logits, y):
    # MSE intre probabilitatile softmax si vectorii one-hot ai etichetelor
    return F.mse_loss(F.softmax(logits, dim=1), F.one_hot(y, output_size).float())

functii_cost = {
    "cross_entropy": nn.CrossEntropyLoss(),
    "mse": mse_loss,
    "hinge": nn.MultiMarginLoss(),
}
criterion = functii_cost[FUNCTIE_COST]

# Regularizarea L2 este implementata prin parametrul weight_decay al optimizatorului
weight_decay = 1e-3 if REGULARIZARE == "l2" else 0.0

# Definim optimizatorul
if OPTIMIZATOR == "sgd":
    optimizer = torch.optim.SGD(model.parameters(), lr=LEARNING_RATE, weight_decay=weight_decay)
elif OPTIMIZATOR == "sgd_momentum":
    optimizer = torch.optim.SGD(model.parameters(), lr=LEARNING_RATE, momentum=0.9, weight_decay=weight_decay)
elif OPTIMIZATOR == "adam":
    optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=weight_decay)
```

```python
# O epoca de antrenare: parcurgem toate batch-urile si actualizam ponderile dupa fiecare
def train_epoch(model, loader):
    model.train()  # activeaza dropout-ul (daca exista)
    total_loss, corecte, total = 0.0, 0, 0
    for X_batch, y_batch in loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        optimizer.zero_grad()            # stergem gradientii de la pasul anterior
        logits = model(X_batch)          # propagare inainte
        loss = criterion(logits, y_batch)
        loss.backward()                  # propagare inapoi (calculul gradientilor)
        optimizer.step()                 # actualizarea ponderilor
        total_loss += loss.item() * len(y_batch)
        corecte += (logits.argmax(dim=1) == y_batch).sum().item()
        total += len(y_batch)
    return total_loss / total, corecte / total

# Evaluarea: doar propagare inainte, fara calculul gradientilor si fara actualizarea ponderilor
@torch.no_grad()
def evaluate(model, loader):
    model.eval()  # dezactiveaza dropout-ul
    total_loss, corecte, total = 0.0, 0, 0
    for X_batch, y_batch in loader:
        X_batch, y_batch = X_batch.to(device), y_batch.to(device)
        logits = model(X_batch)
        total_loss += criterion(logits, y_batch).item() * len(y_batch)
        corecte += (logits.argmax(dim=1) == y_batch).sum().item()
        total += len(y_batch)
    return total_loss / total, corecte / total

# Bucla de antrenare / testare
istoric = {"train_loss": [], "train_acc": [], "test_loss": [], "test_acc": []}
for epoca in range(NR_EPOCI):
    train_loss, train_acc = train_epoch(model, train_loader)
    test_loss, test_acc = evaluate(model, test_loader)
    istoric["train_loss"].append(train_loss)
    istoric["train_acc"].append(train_acc)
    istoric["test_loss"].append(test_loss)
    istoric["test_acc"].append(test_acc)
    if (epoca + 1) % 10 == 0 or epoca == 0:
        print("Epoca {:3d}/{}: train loss = {:.4f}, train acc = {:.3f} | test loss = {:.4f}, test acc = {:.3f}".format(
            epoca + 1, NR_EPOCI, train_loss, train_acc, test_loss, test_acc))

print("\nAcuratete finala pe setul de test: {:.3f}".format(istoric["test_acc"][-1]))
```

```python
# Afisam evolutia functiei de cost si a acuratetii pe train si test
epoci = range(1, NR_EPOCI + 1)
fig, (ax_loss, ax_acc) = plt.subplots(1, 2, figsize=(12, 4))

ax_loss.plot(epoci, istoric["train_loss"], color="tab:blue", label="train")
ax_loss.plot(epoci, istoric["test_loss"], color="tab:orange", label="test")
ax_loss.set_xlabel("Epoca")
ax_loss.set_ylabel("Cost ({})".format(FUNCTIE_COST))
ax_loss.set_title("Functia de cost")
ax_loss.legend()

ax_acc.plot(epoci, istoric["train_acc"], color="tab:blue", label="train")
ax_acc.plot(epoci, istoric["test_acc"], color="tab:orange", label="test")
ax_acc.set_xlabel("Epoca")
ax_acc.set_ylabel("Acuratete")
ax_acc.set_ylim(0, 1.05)
ax_acc.set_title("Acuratete")
ax_acc.legend()

fig.suptitle("activare={}, cost={}, lr={}, batch={}, optimizator={}, regularizare={}, standardizare={}".format(
    ACTIVARE, FUNCTIE_COST, LEARNING_RATE, BATCH_SIZE, OPTIMIZATOR, REGULARIZARE, STANDARDIZARE))
fig.tight_layout()
plt.show()
```
